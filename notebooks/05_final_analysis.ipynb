### Step 14: Underfitting & Overfitting Diagnostics
#### Diagnostic Evaluation Criteria:
- **Overfitting (High Variance):** Would occur if the model scored very high on training data (e.g. $R^2 > 0.90$) but degraded significantly on test data (e.g. $R^2 < 0.40$).
- **Underfitting (High Bias):** Would occur if the model failed to capture data relationships on both training and testing sets equally ($R^2 < 0.10$).
- **Observed Finding:** 
  - Regression: Train $R^2 = 0.5449$ vs. Test $R^2 = 0.5480$ (Difference: $0.0031$).
  - Classification: Train Accuracy = $81.84\%$ vs. Test Accuracy = $81.66\%$ (Difference: $0.18\%$).
  - The near-identical performance on training and unseen testing partitions demonstrates high model stability and zero evidence of overfitting.
